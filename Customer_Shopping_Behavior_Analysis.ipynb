```python
## Key Findings

1. Clothing earns the most revenue: $104,264 of $233,081 total (about 45%).
2. Men make about 68% of purchases. The average purchase is almost the same for men and women (about $60), so the gap comes from the number of purchases, not the amount.
3. Customers in their 20s to 60s each bring in about $43K to $47K.
4. Revenue is spread evenly across seasons. Fall is highest and Summer lowest, a gap of under 8%.
5. The top 10 states and top 10 items are all close together, so no single state or product dominates.
6. Discounts do not raise the average purchase ($59.30 with a discount vs $60.10 without).
7. Subscribers are about 27% of purchases and spend the same as non-subscribers.
8. Payment methods and shipping types are used about equally.
```

```python
## Recommendations

1. Focus on Clothing and Accessories, which bring in about 77% of revenue.
2. Run campaigns aimed at women, since men make most of the purchases.
3. Review the discount strategy. Discounts did not raise the average purchase, so test smaller or targeted offers.
4. Grow subscriptions. Only about 27% of purchases come from subscribers, so add benefits that make subscribing worth it.
5. Keep all payment and shipping options, since customers use them about equally.
```

```python
## Limitations

1. The data has no dates, so I could not study trends over time, such as monthly sales growth.
2. Spending is very similar across almost every group (about $60 per purchase), which suggests the data may be synthetic (made up), so the findings describe this sample only.
3. The original file had two payment columns that disagreed with each other, so I used the payment_method column and noted the choice.
```
